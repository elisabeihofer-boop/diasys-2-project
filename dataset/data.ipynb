```python
# saving dataset with selected columns to json file

dataset.to_json(
    "idioms.json",
    orient="records",
    indent=4,
    force_ascii=False
)
```

```python
# grouping multiple instances of the same idiom together !!!!

grouped_idioms = []

group_columns = [
    "idiom_canonical",
    "idiom_canonical_meaning"
]

for (idiom, meaning), group in dataset.groupby(
    group_columns,
    sort=False,
    dropna=False
):
    first = group.iloc[0]

    examples = []
    seen = set()

    for _, row in group.iterrows():
        sentence = row.get("example")

        if pd.isna(sentence) or not str(sentence).strip():
            continue

        sentence = str(sentence).strip()

        if sentence in seen:
            continue

        seen.add(sentence)

        examples.append({
            "sentence": sentence,
            "meaning": row.get(
                "idiom_in_example_meaning_en", ""
            ),
            "explanation": row.get(
                "explanation_en", ""
            ),
            "context_type": row.get(
                "context_type", ""
            ),
            "tags": row.get("tags", ""),
            "is_generated": row.get(
                "is_generated_example", False
            ),
            "validation_status": row.get(
                "validation_status", ""
            )
        })

    grouped_idioms.append({
        "idiom": idiom,
        "meaning": meaning,
        "idiom_explanation": first.get(
            "idiom_level_explanation_en", ""
        ),
        "register": first.get(
            "idiom_register", ""
        ),
        "difficulty": first.get(
            "learner_difficulty", ""
        ),
        "examples": examples
    })
```

```python
# fixing error I got later when adding to Qdrant

def replace_nan(value):
    if isinstance(value, dict):
        return {k: replace_nan(v) for k, v in value.items()}
    elif isinstance(value, list):
        return [replace_nan(v) for v in value]
    elif pd.api.types.is_scalar(value) and pd.isna(value):
        return None
    return value

grouped_idioms = replace_nan(grouped_idioms)
```

```python
# saving as JSON file

with open("idioms_grouped.json", "w", encoding="utf-8") as f:
    json.dump(
        grouped_idioms,
        f,
        ensure_ascii=False,
        indent=2,
        allow_nan=False
    )

print(f"Created {len(grouped_idioms)} grouped records.")
```
